## 1) What is `print()`?
`print()` is used to **display output** on the screen (console).

```python
print("Hello, World!")
```

## 2) Printing Text and Numbers

```python
print("My name is Ali")
print("Age:", 20)
print(10)
print(10 + 5)
```

## 3) Printing Multiple Values
By default, Python puts a **space** between values.

```python
print("Python", "is", "easy")
print("Marks:", 92, "Grade:", "A")
```

## 4) `sep` (Separator)
Use `sep` to change the separator between printed values.

```python
print("Python", "AI", "Course", sep=" | ")
print("2025", "12", "30", sep="-")
```

## 5) `end` (Ending)
By default, `print()` ends with a new line. Use `end` to change it.

```python
print("Hello", end=" ")
print("Students!")

print("Loading", end="...")
print("Done!")
```

## 6) New Line and Tab (Escape Characters)
- `\n` = new line
- `\t` = tab

```python
print("Line 1\nLine 2\nLine 3")
print("Name\tScore")
print("Ali\t95")
print("Sara\t88")
```
