## 3. Accessing Elements

```python
fruits = ['apple', 'banana', 'cherry']
fruits[0], fruits[-1]
```

## 4. Slicing Lists

```python
nums = [0,1,2,3,4,5]
nums[1:4], nums[:3], nums[::2]
```

## 5. Modifying Lists

```python
fruits.append('orange')
fruits.insert(1,'mango')
fruits.extend(['kiwi','melon'])
fruits
```

## 6. Removing Elements

```python
fruits.remove('banana')
removed = fruits.pop()
fruits, removed
```

## 7. List Functions and Methods

```python
lst = [1,2,2,3]
lst.append(4)
lst.insert(1,10)
lst.extend([5,6])
lst.remove(2)
lst.count(2), lst.index(3)
```

```python
lst.sort()
lst.reverse()
copy_lst = lst.copy()
len(lst), sum(lst), min(lst), max(lst), copy_lst
```

## 8. Lists with Loops

```python
for item in lst:
    print(item)
```

## 9. Lists with String Functions

```python
names = ['ali','ahmed','sara']
[name.upper() for name in names]
```

## 10. List Comprehension

```python
squares = [x*x for x in range(1,6)]
evens = [x for x in range(1,21) if x%2==0]
squares, evens
```

## 11. Nested Lists

```python
matrix = [[1,2,3],[4,5,6]]
for row in matrix:
    for item in row:
        print(item, end=' ')
    print()
```

## 13. Practice Exercises

1. Find unique elements in a list

```python
elements=[1,1,1,2,2,3,4,4,4]
low_occur=elements[0]
lower=elements.count(low_occur)
unique=lower

for element in elements:
    temp=elements.count(element)
    if temp<lower:
        lower=temp
        unique=element

print(unique)

        
```

2. Longest and shortest word

```python
sentence=["happ", "love", "python"]
shortest=sentence[0]
longest=sentence[0]

for word in sentence:
    if len(word)>len(longest):
        longest=word
    else:
        shortest=word


print(shortest,longest)
    
    
    
    
```
